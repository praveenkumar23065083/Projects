```python
class PHYULSTM:

  def __init__(self, eta, eta_t, g, ag, lift, Phi_t):
      # data
      self.eta = eta
      self.eta_t = eta_t
      # self.eta_tt = eta_tt
      self.g = g
      self.ag = ag
      self.lift = lift
        # self.ag_c = ag_c
      self.Phi_t = Phi_t
        # self.Phi_tt = Phi_tt
      self.sess = tf.compat.v1.Session(config=tf.compat.v1.ConfigProto(allow_soft_placement=True,
                                                                  log_device_placement=True))
      self.learning_rate = tf.compat.v1.placeholder(tf.float32, shape=[])
      self.eta_tf = tf.compat.v1.placeholder(tf.float32, shape=[None, None, self.eta.shape[2]])
      self.eta_t_tf = tf.compat.v1.placeholder(tf.float32, shape=[None, None, self.eta.shape[2]])
      #self.eta_tt_tf = tf.compat.v1.placeholder(tf.float32, shape=[None, None, 1])
      self.g_tf = tf.compat.v1.placeholder(tf.float32, shape=[None, None, self.g.shape[2]])
      self.ag_tf = tf.compat.v1.placeholder(tf.float32, shape=[None, None, 1])
      self.lift_tf = tf.compat.v1.placeholder(tf.float32, shape=[None, None, 1])
      self.eta_pred, self.eta_t_pred, self.eta_dot_pred, self.g_pred, self.lift_pred=self.net_structure(self.ag_tf)
        #loss
      self.loss_u = tf.reduce_mean(tf.square(self.eta_tf - self.eta_pred))
      self.loss_udot = tf.reduce_mean(tf.square(self.eta_t_tf - self.eta_dot_pred))
      self.loss_ut = tf.reduce_mean(tf.square(self.eta_dot_pred - self.eta_t_pred))
      self.loss_g = tf.reduce_mean(tf.square(self.g_tf - self.g_pred))
      self.loss_f = tf.reduce_mean(tf.square(self.lift_tf - self.lift_pred))
      self.loss = self.loss_u + self.loss_ut + self.loss_udot + self.loss_g + self.loss_f
   # optimizers
      self.optimizer = tf.compat.v1.train.AdamOptimizer(learning_rate=self.learning_rate)
      self.train_op_Adam = self.optimizer.minimize(self.loss)

      self.optimizer = dde.optimizers.tensorflow_compat_v1.scipy_optimizer.ScipyOptimizerInterface(self.loss,
                                                          method = 'L-BFGS-B',
                                                          options = {'maxiter': 50000,
                                                                      'maxfun': 50000,
                                                                      'maxcor': 50,
                                                                      'maxls': 50,
                                                                      'ftol' : 1.0 * np.finfo(float).eps})
      self.sess = tf.compat.v1.Session(config=tf.compat.v1.ConfigProto(allow_soft_placement=True,
                                                        log_device_placement=True))

      init = tf.compat.v1.global_variables_initializer()
      self.sess.run(init)
# Define convolutional block
  def conv_block(self, input, num_filters):
      x = Conv1D(num_filters, 2, padding="causal")(input)
      x = BatchNormalization()(x)
      x = Activation("relu")(x)
      x = Conv1D(num_filters, 2, padding="causal")(x)
      x = BatchNormalization()(x)
      x = Activation("relu")(x)
      return x

# Define encoder block
  def encoder_block(self, input, num_filters):
      x = self.conv_block(input, num_filters)
      p = MaxPooling1D(2, padding="same")(x)
      return x, p

# Define decoder block
  def decoder_block(self, input, skip_features, num_filters):
    x = UpSampling1D(2)(input)
    #x=input # my modificatn
    #print(f"Shape after upsampling: {x.shape}")
    print(f"Shape of skip_features: {skip_features.shape}")
   # if x.shape[1] != skip_features.shape[1]:
            #x = tf.keras.layers.Cropping1D(cropping=(0, x.shape[1] - skip_features.shape[1]))(x)
   # print(f"Shape after cropping: {x.shape}")
    x = Concatenate()([x, skip_features])
    x = Conv1D(num_filters, 2, padding="causal")(x)
    x = BatchNormalization()(x)
    x = Activation("relu")(x)
    x = Conv1D(num_filters, 2, padding="causal")(x)
    x = BatchNormalization()(x)
    x = Activation("relu")(x)
    return x

# Build U-Net model
  def UNet(self, inputs):
      #inputs = Input(shape=(None, 1, 1))
            # Define a ZeroPadding1D layer with padding at the beginning
      #padding_amount_before = 39
      padding_amount_before = 39
      padding_amount_after=20
      padding_amount = padding_amount_before + padding_amount_after
        # Number of zeros to add before the data
      #zero_padding_layer = ZeroPadding1D(padding=(padding_amount, 0))
      inputs1= ZeroPadding1D(padding=(padding_amount_before, 0))(inputs)
      inputs2= ZeroPadding1D(padding=(0,padding_amount_after))(inputs1)
    # Encoder
      s1, p1 = self.encoder_block(inputs2, 50)
      s2, p2 = self.encoder_block(p1, 100)
      #s3, p3 = self.encoder_block(p2, 200)


    # Bridge
      b1 = self.conv_block(p2, 200)

    # Decoder
      d1 = self.decoder_block(b1, s2, 100)
      d2 = self.decoder_block(d1, s1, 50)
      #d3 = self.decoder_block(d2, s1, 50)
      #d4 = self.decoder_block(d3, s1, 50)
#########################################################################
    # Output
      outputs = Conv1D(3, 1, padding="same", activation="tanh")(d2) ## previously it was sigmoid 
      final_output = Conv1D(3* self.eta.shape[2], 1 , activation="linear")(outputs)
      return final_output

# Define input shape for 1D data
  #input_shape = (None, 1, 1)
  #unet_model = build_UNet(input_shape)

# Assuming self.eta_tt.shape[2] is the number of output channels
  #output_channels = self.eta_tt.shape[2]
  #output_layer = Conv2D(output_channels, (1, 1), activation="linear")(unet_model.output)

# Create the final model
  #final_model = Model(inputs=unet_model.input, outputs=output_layer)

# Usage
# X is your input tensor
  #y = final_model(X)


  def LSTM_model(self, X):
      model = Sequential()
      model.add(LSTM(100, return_sequences=True, stateful=False, input_shape=(None, 3)))
      model.add(Activation('relu'))
      model.add(LSTM(100, return_sequences=True, stateful=False))
      model.add(Activation('relu'))
      model.add(LSTM(100, return_sequences=True, stateful=False))
      model.add(Activation('relu'))
      model.add(Dense(100))
      model.add(Dense(3 * self.eta.shape[2]))
      y = model(X)
      return y
  def net_structure(self, ag):

      output = self.LSTM_model(ag)

      eta     = output[:, :, 0:1]
      eta_dot = output[:, :, 1:2]
      g       = output[:, :, 2:3]

      Phi_ut = np.reshape(
          self.Phi_t,
          [1, self.eta.shape[1], self.eta.shape[1]]
      )

      Phi_ut = np.repeat(
          Phi_ut,
          self.eta.shape[0],
          axis=0
      )

      eta_t = tf.matmul(
          tf.cast(Phi_ut, tf.float32),
          eta
      )

      eta_tt = tf.matmul(
          tf.cast(Phi_ut, tf.float32),
          eta_dot
      )

      lift = eta_tt + g

      return eta, eta_t, eta_dot, g, lift
  def train(self, num_epochs, batch_size, learning_rate, bfgs):
      Loss_u = []
      Loss_udot = []
      Loss_ut = []
      Loss_g = []
      Loss_f = []
      Loss = []

      for epoch in range(num_epochs):

          N = self.eta.shape[0]

          start_time = time.time()
          for it in range(0, N, batch_size):
              tf_dict = {self.eta_tf: self.eta, self.eta_t_tf: self.eta_t, self.g_tf: self.g,
                           self.ag_tf: self.ag, self.lift_tf: self.lift, self.learning_rate: learning_rate}

              self.sess.run(self.train_op_Adam, tf_dict)

                # Print
              if it % (10*batch_size) == 0:
                  elapsed = time.time() - start_time
                  loss_value, learning_rate_value = self.sess.run([self.loss, self.learning_rate], tf_dict)
                  print('Epoch: %d, It: %d, Loss: %.3e, Time: %.2f, Learning Rate: %.3e'
                        %(epoch, it/batch_size, loss_value, elapsed, learning_rate_value))
                  start_time = time.time()
          Loss_u.append(self.sess.run(self.loss_u, tf_dict))
          Loss_ut.append(self.sess.run(self.loss_ut, tf_dict))
          Loss_udot.append(self.sess.run(self.loss_udot, tf_dict))
          Loss_g.append(self.sess.run(self.loss_g, tf_dict))
          Loss_f.append(self.sess.run(self.loss_f, tf_dict))
          Loss.append(self.sess.run(self.loss, tf_dict))



      if bfgs == 1:
         tf_dict_all ={self.eta_tf: self.eta, self.eta_t_tf: self.eta_t, self.g_tf: self.g,
                       self.ag_tf: self.ag, self.lift_tf: self.lift, self.learning_rate: learning_rate}

         self.optimizer.minimize(self.sess,
                                 feed_dict=tf_dict_all,
                                 fetches=[self.loss],
                                 loss_callback=self.callback)
         Loss_u.append(self.sess.run(self.loss_u, tf_dict))
         Loss_ut.append(self.sess.run(self.loss_ut, tf_dict))
         Loss_udot.append(self.sess.run(self.loss_udot, tf_dict))
         Loss_g.append(self.sess.run(self.loss_g, tf_dict))
         Loss_f.append(self.sess.run(self.loss_f, tf_dict))
         Loss.append(self.sess.run(self.loss, tf_dict))


      return Loss_u,Loss_udot, Loss_ut, Loss_g, Loss_f, Loss


  def callback(self, loss):
      print('Loss:', loss)

  def predict(self, ag_star):

      tf_dict = {self.ag_tf: ag_star}

      eta_star = self.sess.run(self.eta_pred, tf_dict)
      eta_t_star = self.sess.run(self.eta_t_pred, tf_dict)
      eta_dot_star = self.sess.run(self.eta_dot_pred, tf_dict)
      g_star = self.sess.run(self.g_pred, tf_dict)

      return eta_star, eta_t_star, eta_dot_star, g_star
```

```python
import numpy as np

# Extract arrays
y_train_ref = np.squeeze(data['y_train_ref'])
y_train_pred = np.squeeze(data['y_train_pred'])

y_pred_ref = np.squeeze(data['y_pred_ref'])
y_pred = np.squeeze(data['y_pred'])

# Correlation arrays
train_correlations = np.zeros(y_train_ref.shape[0])
test_correlations = np.zeros(y_pred_ref.shape[0])

# Training correlations
for i in range(y_train_ref.shape[0]):

    true_signal = y_train_ref[i].flatten()
    pred_signal = y_train_pred[i].flatten()

    if np.std(true_signal) == 0 or np.std(pred_signal) == 0:
        train_correlations[i] = np.nan
    else:
        train_correlations[i] = np.corrcoef(
            true_signal,
            pred_signal
        )[0, 1]

# Test correlations
for i in range(y_pred_ref.shape[0]):

    true_signal = y_pred_ref[i].flatten()
    pred_signal = y_pred[i].flatten()

    if np.std(true_signal) == 0 or np.std(pred_signal) == 0:
        test_correlations[i] = np.nan
    else:
        test_correlations[i] = np.corrcoef(
            true_signal,
            pred_signal
        )[0, 1]

print("Training correlations:")
print(train_correlations)

print("\nTest correlations:")
print(test_correlations)

print("\nSummary:")
print(f"Training Mean PCC = {np.nanmean(train_correlations):.4f}")
print(f"Training Std  PCC = {np.nanstd(train_correlations):.4f}")

print(f"Test Mean PCC = {np.nanmean(test_correlations):.4f}")
print(f"Test Std  PCC = {np.nanstd(test_correlations):.4f}")
```

```python
import scipy.io
import numpy as np
import matplotlib.pyplot as plt



# Extract
y_pred_ref = np.squeeze(data['y_pred_ref'])
y_pred     = np.squeeze(data['y_pred'])

yt_pred_ref = np.squeeze(data['yt_pred_ref'])
yt_pred     = np.squeeze(data['yt_pred'])

g_pred_ref = np.squeeze(data['g_pred_ref'])
g_pred     = np.squeeze(data['g_pred'])

# --------------------------------------------------
# Pearson correlations for each test earthquake
# --------------------------------------------------

disp_corr = []
vel_corr = []
acc_corr = []

for i in range(y_pred_ref.shape[0]):

    # displacement
    r = np.corrcoef(
        y_pred_ref[i].flatten(),
        y_pred[i].flatten()
    )[0,1]
    disp_corr.append(r)

    # velocity
    r = np.corrcoef(
        yt_pred_ref[i].flatten(),
        yt_pred[i].flatten()
    )[0,1]
    vel_corr.append(r)

    # acceleration
    r = np.corrcoef(
        g_pred_ref[i].flatten(),
        g_pred[i].flatten()
    )[0,1]
    acc_corr.append(r)

disp_corr = np.array(disp_corr)
vel_corr = np.array(vel_corr)
acc_corr = np.array(acc_corr)

# --------------------------------------------------
# Violin plot
# --------------------------------------------------

fig, ax = plt.subplots(figsize=(8,6))

parts = ax.violinplot(
    [disp_corr, vel_corr, acc_corr],
    showmeans=True,
    showmedians=True
)

ax.set_xticks([1,2,3])
ax.set_xticklabels([
    'Displacement',
    'Velocity',
    'Acceleration'
])

ax.set_ylabel('Pearson Correlation Coefficient')
ax.set_ylim([-1,1])

ax.grid(True, alpha=0.3)

print("Displacement")
print("Mean =", np.mean(disp_corr))
print("Std  =", np.std(disp_corr))

print("\nVelocity")
print("Mean =", np.mean(vel_corr))
print("Std  =", np.std(vel_corr))

print("\nAcceleration")
print("Mean =", np.mean(acc_corr))
print("Std  =", np.std(acc_corr))

plt.tight_layout()
plt.show()
```
